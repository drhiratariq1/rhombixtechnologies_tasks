```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Check the number of rows and columns

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
# Display all column names

print("Dataset Columns:")
print(df.columns.tolist())
```

```python
# Understand dataset structure

df.info()
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
display(df.head())
display(df.tail())
```

```python
# Check missing values in each column

print(df.isnull().sum())
```

```python
# Calculate percentage of missing values

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
# Identify duplicate records

print("Total duplicate records:", df.duplicated().sum())
```

```python
# Check data types of all columns

print(df.dtypes)
```

```python
# Examine categorical variables

categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    print("\nColumn:", col)
    print(df[col].unique())
```

```python
# Check for extra spaces in categorical columns

for col in categorical_columns:
    print(col, ":", df[col].str.strip().equals(df[col]))
```

```python
# Final data quality report

print("FINAL DATA QUALITY REPORT")
print("-------------------------")

print("Dataset Shape:", df.shape)

print("Total Missing Values:", df.isnull().sum().sum())

print("Total Duplicate Rows:", df.duplicated().sum())

print("Total Columns:", len(df.columns))

print("Total Rows:", len(df))

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(df.head())
```

df.shape: Checks the number of rows and columns.

df.isnull().sum().sum(): Counts all missing values.

df.duplicated().sum(): Counts duplicate records.

len(df.columns): Counts columns.

len(df): Counts rows.

df.dtypes: Displays column data types.

df.head(): Displays the first five records.

```python
# Export cleaned dataset to CSV

df.to_csv("Cleaned_StudentsPerformance.csv", index=False)

print("Cleaned dataset exported successfully!")
```

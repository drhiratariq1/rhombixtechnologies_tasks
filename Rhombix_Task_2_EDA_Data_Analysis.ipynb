```python

for column in ["math score", "reading score", "writing score"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\n{column}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

This code calculates:
- Q1: First quartile.
- Q3: Third quartile.
- IQR: Difference between Q3 and Q1.
- Lower and upper limits: Boundaries used to identify potential outliers.
- Number of outliers: Values outside those boundaries.

### Interpretation of Outlier Analysis Using IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in mathematics, reading, and writing scores.

**Results:**

- Mathematics has 8 potential outliers.
- Reading has 6 potential outliers.
- Writing has 5 potential outliers.

**Interpretation:** Mathematics has the highest number of potential outliers (8), followed by reading (6) and writing (5). All three subjects have upper limits greater than 100, while their lower limits are below 30. Since the maximum score is 100, the identified potential outliers are unusually low scores.

These observations will be investigated further before deciding whether any data cleaning is necessary.

```python

for column in ["math score", "reading score", "writing score"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df.loc[
        (df[column] < lower_limit) |
        (df[column] > upper_limit), column
    ]

    print("\n", column)
    print("Outlier values:", sorted(outliers.tolist()))
```

### Identification of Actual Outlier Values

The IQR method identified the following potential outlier values:

- Mathematics: 0, 8, 18, 19, 22, 23, 24, and 26.
- Reading: 17, 23, 24, 24, 26, and 28.
- Writing: 10, 15, 19, 22, and 23.

**Interpretation:** These values represent unusually low student scores compared with the overall score distributions. Mathematics contains the lowest potential outlier (0). However, these values may represent genuine student performance rather than data errors.

**Conclusion:** All identified potential outliers will be retained because there is no evidence that they are incorrect observations. Removing genuine student scores could affect the accuracy of the analysis.

### Average Score Analysis

```python
bins = [0, 50, 70, 85, 100]

labels = ['Low', 'Average', 'Good', 'Excellent']

df['performance level'] = pd.cut(
    df['average score'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df[['average score', 'performance level']].head(10)
```

```python
df['performance level'].value_counts().sort_index()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='performance level',
    order=labels
)

plt.title('Distribution of Students by Performance Level')
plt.xlabel('Performance Level')
plt.ylabel('Number of Students')

plt.tight_layout()
plt.show()
```
